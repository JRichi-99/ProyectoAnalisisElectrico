```python
import pandas as pd
import zipfile 
from pathlib import Path 
```

```python
db_path = Path(r"D:\BDDPablacomCoordinador")
```

```python
target_bdd_new = ["Medidas_Valorizadas_15min_Norte Distribución",
                    "Medidas_Valorizadas_15min_Norte",
                    "Medidas_Valorizadas_15min_Sur Distribución",
                    "Medidas_Valorizadas_15min_Sur"]
```

Extraemos los zips para no hacerlo uno por uno

```python
for folder_date in db_path.iterdir():
    if not folder_date.is_dir():
        continue
        
    origin, date, version = folder_date.name.split("_")
    
    if int(date) < 2505:
        # Base de datos antigua
        folder_medidas = folder_date / "04 Valorizado"
        lista_medidas = ["Balance_Valorizado_{}_Data_VALORIZADO_15min".format(date)]
    else:
        # Base de datos nueva
        folder_medidas = folder_date / "02 Medidas por tipo"
        lista_medidas = target_bdd_new

    for name_medida in lista_medidas:
        path_zip_medida = folder_medidas / "{}.zip".format(name_medida)
        path_folder_medida = folder_medidas / name_medida
        
        if path_folder_medida.exists() or not path_zip_medida.exists():
            continue
            
        try:
            with zipfile.ZipFile(path_zip_medida, "r") as zip_ref:
                zip_ref.extractall(path_folder_medida)
                print("Extracción completa de {}.{} en {}".format(date,
                                                                  path_zip_medida.name, 
                                                                  path_folder_medida.name))
            path_zip_medida.unlink(missing_ok=True)
            print("Eliminacion del zip realizada")
            
        except:
            print(f"No se logró con el archivo bajo {folder_date.name}")
            continue
```
