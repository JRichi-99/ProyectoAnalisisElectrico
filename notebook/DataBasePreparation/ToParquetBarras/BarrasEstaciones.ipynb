```python
import pandas as pd
import numpy as np
from pathlib import Path
```

```python
macrozonas = {
    # 1.- Macrozona Norte
    'Arica y Parinacota': 'Norte Grande',
    'Tarapacá': 'Norte Grande',
    'Antofagasta': 'Norte Grande',
    
    'Atacama': 'Norte Chico',
    'Coquimbo': 'Norte Chico',
    
    'Valparaíso': 'Centro',
    'Metropolitana de Santiago': 'Centro', # En realidad esta es una aparte pero creo que es suficiente si es agrupada
    
    # 4.- Macrozona Centro Sur
    'Libertador Gral. Bernardo O\'Higgins': 'Centro Sur',
    'Maule': 'Centro Sur',
    'Ñuble': 'Centro Sur',
    'Biobío': 'Centro Sur',
    
    # 5.- Macrozona Sur
    'La Araucanía': 'Sur',
    'Los Ríos': 'Sur',
    'Los Lagos': 'Sur',
    
    # 6.- Macrozona Austral
    'Aysén del General Carlos Ibáñez del Campo': 'Austral',
    'Magallanes y de la Antártica Chilena': 'Austral'
}
```

```python
df_subestaciones = pd.read_excel(Path(r"D:\ProyectoAnalisisElectrico\BarrasEstaciones\reporte_subestaciones.xlsx"),
                                 skiprows=6)
df_barras = pd.read_excel(Path(r"D:\ProyectoAnalisisElectrico\BarrasEstaciones\reporte_barras.xlsx"),
                          skiprows=6)
```

```python

df = pd.merge(left = df_subestaciones, right = df_barras, left_on="Nombre", right_on="Nombre Subestación", how="outer", suffixes=("_SE", "_BA"))
```

```python
df.columns
```

```python
df = df[['ID_SE', 
         'Nombre_SE',
         'Nombre Coordinado_SE',
         'Nemotecnico_SE', 
         'Región', 
         'Provincia', 
         'Comuna',
         'Nombre_BA', 
         'Nemotecnico_BA'
         ]]
df.head()
```

```python
filas_na_por_barra = len(df[df["Nombre_BA"].isna()])
filas_na_por_estaciones =  len(df[df["Nombre_SE"].isna()])
print(f"A {filas_na_por_barra} estaciones no se les encontro barras de {len(df_subestaciones)}")
print(f"A {filas_na_por_estaciones} barras no se les encontro estacion de {len(df_barras)}")
```

```python
df["Macrozona"] = df["Región"].map(macrozonas)
```

```python
to_save_path = Path(r"D:\ProyectoAnalisisElectrico\BarrasEstaciones") / f"BarrasEstaciones.parquet"
if not to_save_path.exists():
    df.to_parquet(to_save_path, engine="pyarrow", compression="snappy")
```
