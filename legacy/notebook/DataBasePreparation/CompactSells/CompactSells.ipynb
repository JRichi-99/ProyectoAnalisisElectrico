```python
import pandas as pd
from pathlib import Path
```

```python
ventas2026 = pd.read_excel(Path(r"E:\ProyectoAnalisisElectrico\TransferenciasEconomicas\Ventas_SEN_2026_v2.xlsx"), sheet_name = "DATA")
ventas2025 = pd.read_excel(Path(r"E:\ProyectoAnalisisElectrico\TransferenciasEconomicas\Ventas_SEN_2025-5.xlsx"), sheet_name = "DATA")
```

```python
ventas2025 = ventas2025[ventas2025["MES"].isin([i for i in range(5, 13)])]
ventas2026 = ventas2026[ventas2026["MES"].isin([i for i in range(1, 5)])]
```

```python
ventas = pd.concat([ventas2026, ventas2025], ignore_index=True)
```

```python
ventas = ventas[ventas["TIPO"].isin(["L", "L_D"])]
ventas = ventas[~ventas["clave"].isna()]
ventas = ventas[~ventas["RUTCLIENTE"].isna()]
```

```python
ventas.to_csv(Path(r"E:\ProyectoAnalisisElectrico\TransferenciasEconomicas\Ventas_SEN_2505_2604.csv"), index=False)
```
