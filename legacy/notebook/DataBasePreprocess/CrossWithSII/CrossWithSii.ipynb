```python
import pandas as pd
import numpy as np
from pathlib import Path

sii_path = Path(r"D:\ProyectoAnalisisElectrico\SII\ActividadesEconomicasMayo2026\PUB_NOM_ACTECOS.txt")
sii = pd.read_csv(sii_path,sep="\t")

carpeta_datos = Path(r"D:\ProyectoAnalisisElectrico\DiaPromedio\Periodos\2505_2604")
print("Cargando versión Porcentual...")
df = pd.read_parquet(carpeta_datos / "2505_2604_clustered_porcentual.parquet")
```

```python
df[["NDV", "DV"]] = df["RUT"].str.replace(".", "", regex=False).str.split("-", expand=True)
df["NDV"] = df["NDV"].astype(int)
```

```python
df1 = pd.merge(left=df, right=sii, how="left", left_on=["NDV","DV"], right_on=["RUT","DV"])
```
